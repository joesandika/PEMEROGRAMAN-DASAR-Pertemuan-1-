```python
# Jawaban Latihan 2
jumlah_led = 5           # int (bilangan bulat)
panjang_kabel = 1.5      # float (bilangan desimal)
lampu_on = True          # bool (True / False)
nama_alat = "Multimeter" # str (teks)

# Fungsi type() untuk memeriksa tipe data
print(jumlah_led, type(jumlah_led))
print(panjang_kabel, type(panjang_kabel))
print(lampu_on, type(lampu_on))
print(nama_alat, type(nama_alat))
```

```python
# Jawaban Latihan 3
nama_unit = "PLTS Nunukan" # str (teks deret)
jumlah_panel = 240         # int (bilangan bulat)
tegangan_keluaran = 48.5   # float (bilangan desimal)
sedang_beroperasi = True   # bool (True / False)

# Buktikan dengan type()
print("nama_unit        :", nama_unit, "| Tipe:", type(nama_unit))
print("jumlah_panel     :", jumlah_panel, "| Tipe:", type(jumlah_panel))
print("tegangan_keluaran:", tegangan_keluaran, "| Tipe:", type(tegangan_keluaran))
print("sedang_beroperasi:", sedang_beroperasi, "| Tipe:", type(sedang_beroperasi))
```

```python
# Jawaban Latihan 4
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Cetak pembacaan pertama dan terakhir memakai indeks
print("Pembacaan Pertama:", tegangan_harian[0], "V")
print("Pembacaan Terakhir:", tegangan_harian[-1], "V")

# (b) Tambahkan satu pembacaan baru 222.0 dengan append()
tegangan_harian.append(222.0)
print("Setelah_ditambah :", tegangan_harian)

# (c) Cetak jumlah total data dengan len()
total_data = len(tegangan_harian)
print("Jumlah Total Data :", total_data)
```

```python
# Jawaban Latihan 5
motor_listrik = {
    "nama": "Motor Induksi 3 Fasa",
    "daya_kW": 5.5,
    "tegangan_V": 380,
    "rpm" : 1450
}
# (a) Cetak salah satu nilai lewat key-nya
print("Tegangan Motor:", motor_listrik["tegangan_V"], "V")

# (b) Mengubah nilai rpm
motor_listrik["rpm"] = 1430

# (c) Cetak ulang seluruh dictionary.
print("\nSpesisikasi Terbaru Motor Listrik:")
print(motor_listrik)
```

```python
# Jawaban Latihan 6
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80,  "status": "pemeliharaan"},
]

# cetak nama unit kedua
print("Nama unit kedua:", pembangkit[1]["nama"])

# Ubah status salah satu unit
pembangkit[0]["status"] = "pemeliharaan"

# Tambah satu unit baru
pembangkit.append({"nama": "Unit-4", "kapasitas_MW": 120, "status": "operasi"})

# Cetak seluruh data
print("\nData Pembangkit:")
print(pembangkit)
```
